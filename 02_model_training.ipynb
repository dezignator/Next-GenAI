# Next-GenAI: Data Exploration
This notebook explores the `doh_real_world_2022` dataset and visualizes the PCA/t-SNE reduction (Sec. 3.1).

```python
import pandas as pd
import matplotlib.pyplot as plt
from src.data_acquisition.preprocessing import PreprocessingPipeline

# Load data
df = pd.read_parquet('../data/raw/doh_real_world_2022.parquet')
print(f"Total rows: {len(df)}")
print(df['label'].value_counts())
```

```python
# Run pipeline and get t-SNE embeddings
pipeline = PreprocessingPipeline()
out = pipeline.run(df, do_tsne=True)

# Plot t-SNE
plt.figure(figsize=(10, 6))
plt.scatter(out['X_tsne'][:, 0], out['X_tsne'][:, 1], c=out['y_train'], alpha=0.5, cmap='coolwarm')
plt.title('t-SNE Projection of DoH Features')
plt.show()
```
